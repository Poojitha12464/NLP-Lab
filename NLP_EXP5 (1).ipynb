```python
# EXP 5: GPT-2 Text Generation Fine-Tuned


from transformers import GPT2LMHeadModel, GPT2Tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
input_text = "Once upon a time"
input_ids = tokenizer.encode(input_text, return_tensors="pt")
output = model.generate(
    input_ids,
    max_length=100,
    do_sample=True,
    top_k=50,#top_k limits the number of candidate tokens considered at each step.
    top_p=0.95, #top_p (nucleus sampling) considers tokens whose cumulative probability meets a threshold.
    temperature=0.8 #temperature controls randomness; lower values make output more deterministic
)
generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print(generated_text)
```

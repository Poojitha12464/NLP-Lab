## Rule-Based Named Entity Recognition (NER)

Named Entity Recognition (NER) is a subtask of information extraction that seeks to locate and classify named entities in unstructured text into pre-defined categories such as person names, organizations, locations, medical codes, time expressions, quantities, monetary values, percentages, etc.

**Rule-based NER** relies on handcrafted rules, patterns, and dictionaries to identify and classify entities. These rules can be based on:

*   **Lexicons/Dictionaries:** Lists of known entities (e.g., a list of country names, common person names).
*   **Regular Expressions:** Patterns to match specific entity formats (e.g., phone numbers, dates, email addresses).
*   **Syntactic and Morphological Patterns:** Rules based on grammar, part-of-speech tags, or word endings.

While rule-based systems can be very effective for specific, well-defined domains, they often require significant manual effort to create and maintain rules, and they may not generalize well to different types of text or entities.

```python
# Let's implement a simple rule-based NER system using regular expressions and a small dictionary.

import re

def rule_based_ner(text):
    entities = {
        'PERSON': [],
        'LOCATION': [],
        'ORGANIZATION': [],
        'DATE': []
    }

    # Rule 1: Recognize specific known names (dictionary-based)
    person_names = ['Alice', 'Bob', 'Charlie', 'Dr. Smith']
    for name in person_names:
        if name in text:
            entities['PERSON'].append(name)

    # Rule 2: Recognize common locations (dictionary-based)
    locations = ['New York', 'London', 'Paris', 'France', 'USA']
    for loc in locations:
        if loc in text:
            entities['LOCATION'].append(loc)

    # Rule 3: Recognize organizations (simple heuristic: capitalized words followed by Inc., Corp., Ltd.)
    organization_pattern = r'\b([A-Z][a-z]+\s(?:Inc\.|Corp\.|Ltd\.)|\b[A-Z][A-Za-z0-9&\s]+(?:Company|Group|Bank|Corporation|Institute))\b'
    organizations = re.findall(organization_pattern, text)
    for org in organizations:
        entities['ORGANIZATION'].append(org.strip())

    # Rule 4: Recognize simple dates (e.g., 'YYYY-MM-DD' or 'Month Day, Year')
    date_patterns = [
        r'\b\d{4}-\d{2}-\d{2}\b',
        r'\b(?:January|February|March|April|May|June|July|August|September|October|November|December)\s\d{1,2},\s\d{4}\b'
    ]
    for pattern in date_patterns:
        dates = re.findall(pattern, text)
        entities['DATE'].extend(dates)

    return {k: list(set(v)) for k, v in entities.items() if v} # Remove duplicates and empty lists

# Example Usage:
text1 = "Alice works at Google Inc. in New York. She visited London on 2023-10-26."
text2 = "Dr. Smith will give a presentation for the Microsoft Corporation on February 15, 2024."

print(f"Text 1: {text1}")
print(f"Entities found: {rule_based_ner(text1)}\n")

print(f"Text 2: {text2}")
print(f"Entities found: {rule_based_ner(text2)}\n")
```

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
## tokenization: convert para to sentences to words
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve LookupError
sentences=nltk.sent_tokenize(paragraph)
```

```python
#stemming
stemmer=PorterStemmer()
```

```python
stemmer.stem('goes')
# to see base of a word
```

```python
#lemmatization
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer=WordNetLemmatizer()
```

The `WordNetLemmatizer` often requires a part-of-speech (POS) tag to correctly lemmatize words that can function as different parts of speech. By default, it assumes the word is a noun.

Since 'helping' is a verb, we need to specify `pos='v'` for correct lemmatization:

```python
lemmatizer.lemmatize('helping', pos='v')
```

```python
import re
corpus=[]
for i in range(len(sentences)):
  review=re.sub('[^a-zA-Z]',' ',sentences[i])
  review=review.lower()
  review.split()
  corpus.append(review)
```

```python
#to print new cleaned text
corpus
```

```python
nltk.download('stopwords') # Download stopwords corpus
# apply Stopwords.Lemmatize
#put this cell before vocabulary
import re
corpus=[]
for i in range (len(sentences)):
  review=re.sub('[^a-zA-Z]',' ',sentences[i])
  review=review.lower()
  review=review.split()
  review=[lemmatizer.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
  review=' '.join(review)
  corpus.append(review)
```

```python
#stemming
for i in corpus:
  words=nltk.word_tokenize(i)
  for word in words:
    if word not in set(stopwords.words('english')):
      print(stemmer.stem(word))
```

```python
# lemmatization
for i in corpus:
  words=nltk.word_tokenize(i)
  for word in words:
    if word not in set(stopwords.words('english')):
      print(lemmatizer.lemmatize(word,pos='v'))
```
